```python

# 8. Data Cleaning - Removing units
print("\n Cleaning columns with units")
unit_columns = [
    'total_fat', 'saturated_fat', 'cholesterol', 'sodium',
    'choline', 'fat', 'saturated_fatty_acids', 'monounsaturated_fatty_acids',
    'polyunsaturated_fatty_acids', 'fatty_acids_total_trans', 'alcohol',
    'ash', 'caffeine', 'theobromine', 'water'
]

def clean_column(col):
    return pd.to_numeric(col.str.replace(r'[^0-9.-]', '', regex=True), errors='coerce')

for col in unit_columns:
    if col in df.columns:
        df[col] = clean_column(df[col])
```

```python
# 9. Handling Missing Values
print("\n Handling Missing Values")
for column in df.columns:
    if df[column].isnull().sum() > 0:
        if df[column].dtype == "object":
            df[column].fillna(df[column].mode()[0], inplace=True)
        elif df[column].dtype in ["int64", "float64"]:
            mean_val = df[column].mean()
            median_val = df[column].median()
            if abs(mean_val - median_val) / mean_val < 0.1:
                df[column].fillna(mean_val, inplace=True)
            else:
                df[column].fillna(median_val, inplace=True)
```

# Q2. Classify food items as “High Fat”, “Medium Fat”, or “Low Fat”

```python
df['fat'] = pd.to_numeric(df['fat'], errors='coerce')
fat_q = df['fat'].quantile([0.33, 0.66])

def classify_fat(f):
    if f <= fat_q[0.33]:
        return "Low Fat"
    elif f <= fat_q[0.66]:
        return "Medium Fat"
    else:
        return "High Fat"

df['fat_category'] = df['fat'].apply(classify_fat)
df['fat_category'].value_counts()

```

# Q8. Cluster food items by nutrition

```python
# Step 1: Clean the numeric columns by removing non-numeric characters
import re

def clean_column(col):
    return col.str.replace(r'[^\d\.\-]', '', regex=True)

cols_to_clean = ['calories', 'protein', 'fat', 'carbohydrate', 'fiber', 'sugars', 'sodium']

for col in cols_to_clean:
    df[col] = clean_column(df[col].astype(str))
    df[col] = pd.to_numeric(df[col], errors='coerce')  # convert to numeric

# Step 2: Drop rows with NaNs in selected columns
df_clean = df.dropna(subset=cols_to_clean).copy()

# ✅ Confirm how many rows are left
print("Remaining rows after cleaning:", len(df_clean))

```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Step 4: Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_clean[cols_to_clean])

# Step 5: Apply KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
df_clean['cluster'] = kmeans.fit_predict(X_scaled)

# Step 6: Show cluster counts
print(df_clean['cluster'].value_counts())
```
